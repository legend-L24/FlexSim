## Here we want to generate the initial structure with any number of H2O

```python
from ase.io import read, write
from ase.build import molecule
from flex_sim.grid import get_accessible_positions
from flex_sim.init import farthest_point_sampling_pbc_vectorized
from flex_sim.molecule import add_molecule
from flex_sim.u_md import GeometryOptimization
from mace.calculators import MACECalculator

ciffile = "/users/lyutao/test_mofs/cau10/CAU-10.cif"
modelpath = '/capstor/scratch/cscs/lyutao/ZnMOF_extend_training/modeldevi0_compare/MACE_stagetwo.model'
structure = read(ciffile)
gas = molecule("H2O")

if_optimized = True
num_of_gas = 70
random_seed = 42

fix_indices = list(range(len(structure)))


ret = get_accessible_positions(
    structure=structure,
    grid_spacing=0.2,
    cutoff_distance=2,
    min_interplanar_distance=10.0,
)

selected_positions = farthest_point_sampling_pbc_vectorized(
    points=ret['accessible_pos'],
    atoms=structure,
    n_samples=num_of_gas,
    random_seed=random_seed
)

for idx in range(num_of_gas):
    gas_molecule = add_molecule(gas, rotate=True, translate=selected_positions[idx])
    structure += gas_molecule
    structure.wrap()


initial_structure = structure.copy()
if if_optimized:
    mace_calc = MACECalculator(model_paths=[modelpath], device='cuda')
    final_structure, flag = GeometryOptimization(
        atoms=structure,
        Fmax=0.01,
        stepMax=200,
        calc=mace_calc, 
        save_traj=False,
        outputatoms=True,
        logfile='opt.log',
        fix_indices=fix_indices
    )
    
```

## Do it for certain examples

```python
from pathlib import Path
from flex_sim.init import insert_and_optimize_gas
from ase.io import read, write
metal_list = ['Ag', 'Al', 'Ba', 'Be', 'Bi', 'Ca', 'Cd', 'Co', 'Cr', 'Cu', 'Fe', 'Ga', 'In', 'K', 'Li', 'Mg', 'Mn', 'Na', 'Ni', 'Pb', 'Pd', 'Pt', 'Rb', 'Sc', 'Sn', 'Sr', 'Ti', 'V', 'Y', 'Zn', 'Zr']

modelpath = '/capstor/scratch/cscs/lyutao/ZnMOF_extend_training/modeldevi0_compare/MACE_stagetwo.model'
gas_name = "H2O"
structure_dir = Path("/users/lyutao/test_mofs/")
cif_files = list(structure_dir.rglob("*.cif"))
dest_dir = Path("/users/lyutao/init_water/")
if not dest_dir.exists():
    dest_dir.mkdir(parents=True, exist_ok=True)
```

```python
cif_files
```

```python
## we expect to generate structures with 1~4 molecules per unit cell

count = 0
for ciffile in cif_files:
    atoms = read(ciffile)
    number_of_metal = sum([1 for atom in atoms if atom.symbol in metal_list])
    print(f"Processing {ciffile}, which has {number_of_metal} metal atoms.")  
    
    for loading_per_metal in range(1, 5):
        total_num_of_gas = number_of_metal * loading_per_metal
        if ciffile.stem == 'CAU-10' and loading_per_metal == 4:
            total_num_of_gas = 54  ## to avoid too dense
        print(f"Inserting {total_num_of_gas} {gas_name} molecules into {ciffile.stem}.")
        final_structure = insert_and_optimize_gas(
            ciffile=str(ciffile),
            gas_name=gas_name,
            modelpath=modelpath,
            num_of_gas=total_num_of_gas,
            if_optimized=True,
            random_seed=42,
            grid_spacing=0.2,
            cutoff_distance=1.6,
            min_interplanar_distance=10.0,
            Fmax=0.01,
            stepMax=200,
            device='cuda',
            logfile='opt.log'
        )
        final_folder = dest_dir/f"{count}"
        if not final_folder.exists():
            final_folder.mkdir(parents=True, exist_ok=True)
        write(final_folder/f"{ciffile.stem}_{loading_per_metal}w.cif", final_structure)
        count += 1
        
```

## rescale the cif

```python
from pathlib import Path

from ase.io import read, write
scale = '112'
structure_dir = Path(f"/capstor/scratch/cscs/lyutao/activelearning/ZnMOF_iter2/template/mof5_data_{scale}/")
cif_files = list(structure_dir.rglob("*.cif"))

for ciffile in cif_files:
    atoms = read(ciffile)
    print(f"Processing {ciffile}, which has {len(atoms)} atoms, scale it to {scale}")
    atoms_scaled = atoms.repeat([int(scale[0]), int(scale[1]), int(scale[2])])
    write(ciffile, atoms_scaled)
    print(f"Written the scaled structure to {ciffile}")
```

```python
from pathlib import Path
import shutil
framefile = "/capstor/scratch/cscs/lyutao/cp2kdata_stress/mof385_opt/3/CAU10.cif"
loadingfile0 = "/capstor/scratch/cscs/lyutao/cp2kdata_stress/mof351_loading_val/cau10/CAU10_loading.cif"
loadingfile1 = "/capstor/scratch/cscs/lyutao/cp2kdata_stress/mof351_loading_val/cau10/CAU10_loading.cif"

number = 8

framework_dir = Path("/capstor/scratch/cscs/lyutao/heat/cau10_test0/frame")
loading_dir = Path("/capstor/scratch/cscs/lyutao/heat/cau10_test0/load")
for idx in range(0, number):
    dest_dir = framework_dir / f"{idx}"
    if not dest_dir.exists():
        dest_dir.mkdir(parents=True, exist_ok=True)
    # copy the framefile to dest_dir, use shutil.copy
    shutil.copy(framefile, dest_dir)
    
for idx in range(0, number*2):
    dest_dir = loading_dir / f"{idx}"
    if not dest_dir.exists():
        dest_dir.mkdir(parents=True, exist_ok=True)
    # copy the loadingfile0 and loadingfile1 alternatively
    if idx > 7:
        shutil.copy(loadingfile1, dest_dir)
    else:
        shutil.copy(loadingfile0, dest_dir)
```
